```python
my_str = "Although that way may not be obvious at first unless you're Dutch."

# Without begin, end and step
print("You have sliced:", my_str[:])

# Begin = 0, end using len(), no step
print("You have sliced:", my_str[0:len(my_str)])

# Without begin and end, but using step
print("You have sliced:", my_str[::1])

# With begin, end and step
print("You have sliced:", my_str[0:len(my_str):1])

# Begin & end positive, step negative → empty string
print("You have sliced:", my_str[5:10:-1])

# Output: Atog htwymyntb biu tfrtuls o'eDth
print("You have sliced:", my_str[::2])

# Output: Ahgttam tebo  r lsorDc
print("You have sliced:", my_str[::3])

# Reverse using only step
print("You have sliced:", my_str[::-1])

# Reverse using begin, end, step
print("You have sliced:", my_str[len(my_str)-1::-1])

# Output: .cu ruysen si asovoe o a a athuhl
print("You have sliced:", my_str[::-3])

# Same using begin, end, step
print("You have sliced:", my_str[len(my_str)-1::-3])


# Given slice
print(my_str[10:17:-1])
# Output: '' (empty string)


# Output: yaw ta
print("You have sliced:", my_str[15:9:-1])

# Output: ess you
print("You have sliced:", my_str[58:51:-1])
```

```python
my_str = "Although 8 that way may not be obvious at first unless you're Dutch"

# Count of 't' using find()
count = 0
i = my_str.find('t')
while i != -1:
    count += 1
    i = my_str.find('t', i + 1)
print(count)

# (Optional) using index()
count = 0
try:
    i = my_str.index('t')
    while True:
        count += 1
        i = my_str.index('t', i + 1)
except ValueError:
    pass
print(count)


# Index of '8'
print(my_str.find('8'))
print(my_str.index('8'))


# Outputs of given statements
print(my_str.find('the'))      # substring 'the'
print(my_str.index('the'))     # same as above

print(my_str.find('t', 9, 15)) # search between index 9–14

print(my_str.rfind('u'))       # last occurrence of 'u'
print(my_str.rindex('u'))      # same as above
```

```python
#W A P which applies strip() method if any string, which will be taken from user, starts and ends with space, or applies
#rrstrip() method if that string only ends with space or applies lstrip() method if that string only starts with a space.

#For example:-
#input:- '    Python   '
#output:- 'Python'

#input:- '    Python'
#output:- 'Python'

#input:- 'Python   '
#output:- 'Python'

# Take input from user
# Take input from user
# Take input from user
my_str = input("Enter a string: ")

# Apply conditions
if my_str.startswith(' ') and my_str.endswith(' '):
    result = my_str.strip()
elif my_str.startswith(' '):
    result = my_str.lstrip()
elif my_str.endswith(' '):
    result = my_str.rstrip()
else:
    result = my_str

print("Output:", result)
```

```python
#Take a string from user and check if it is:-
#     1. alphanumeric
#     2. alphabets
#     3. digit
#     4. all letters are in lower case
#     5. all letters are in upper case
#     6. in title case
#     7. a space character
#     8. numeric
#     9. all number elements in string are decimal
# Take input from user
my_str = input("Enter a string: ")

print("1. Is Alphanumeric:", my_str.isalnum())
print("2. Is Alphabet:", my_str.isalpha())
print("3. Is Digit:", my_str.isdigit())
print("4. All letters are lowercase:", my_str.islower())
print("5. All letters are uppercase:", my_str.isupper())
print("6. Is Title Case:", my_str.istitle())
print("7. Is Space Character:", my_str.isspace())
print("8. Is Numeric:", my_str.isnumeric())
print("9. All numbers are decimal:", my_str.isdecimal())
```

```python
#Write a python program to implement replace method. If sub_str is found in my_str then it will replace the first
#occurrence of sub_str with new_str else it will will print sub_str not found:-

#Input:- my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.', sub_str = 'Peck', new_str = 'Pack'
#Output:- 'Peter Piper Picked A Pack Of Pickled Peppers.'

my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.'
sub_str = 'Peck'
new_str = 'Pack'
if sub_str in my_str:
    result = my_str.replace(sub_str, new_str, 1)  # replace only first occurrence
    print(result)
else:
    print("sub_str not found")
```

```python
#Write a python program to find below output (implements rjust and ljust):-

#Input:- 'Peter Piper Picked A Peck Of Pickled Peppers.', sub_str = 'Peck',
#Output:- '*********************Peck********************'

my_str = 'Peter Piper Picked A Peck Of Pickled Peppers.'
sub_str = 'Peck'
# Find index of 'Peck'
i = my_str.index(sub_str)
# Total length should match original string length
result = sub_str.rjust(i + len(sub_str), '*').ljust(len(my_str), '*')
print(result)
```
