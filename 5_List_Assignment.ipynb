```python
#Write a Python program to find the sum of all elements in a list using loop.
#Input:- [10,20,30,40]
#Output:- 100

my_list = [10, 20, 30, 40]
total = 0
for num in my_list:
    total += num
print(total)
```

```python
#Write a Python program to find the largest number from a list using loop.
#Input:- [10,100,2321, 1,200,2]
#Output:- 2321

my_list = [10, 100, 2321, 1, 200, 2]
largest = my_list[0]
for num in my_list:
    if num > largest:
        largest = num

print(largest)
```

```python
#Write a Python program to find the smallest number from a list using loop.
#Input:- [10,100,2321, 1,200,2]
#Output:- 1
my_list = [10, 100, 2321, 1, 200, 2]
smallest = my_list[0]  # assume first element is smallest
for num in my_list:
    if num < smallest:
        smallest = num

print(smallest)

```

```python
#Write a Python program to count the number of strings having length more than 2 and are palindrome in a list using loop.
#Input:- ['ab', 'abc', 'aba', 'xyz', '1991']
#Output:- 2
my_list = ['ab', 'abc', 'aba', 'xyz', '1991']
count = 0
for s in my_list:
    if len(s) > 2 and s == s[::-1]:
        count += 1

print(count)
```

```python
#Write a Python program to sort a list in ascending order using loop.
#Input:- [100,10,1,298,65,483,49876,2,80,9,9213]
#Output:- [1,2,9,10,65,80,100,298,483,9213,49876]

my_list = [100, 10, 1, 298, 65, 483, 49876, 2, 80, 9, 9213]
n = len(my_list)
for i in range(n):
    for j in range(0, n - i - 1):
        if my_list[j] > my_list[j + 1]:
            # swap
            my_list[j], my_list[j + 1] = my_list[j + 1], my_list[j]
print(my_list)
```

```python
#Write a Python program to get a sorted list in increasing order of last element in each tuple in a given list using loop.
#Input:- [(5,4),(9,1),(2,3),(5,9),(7,6),(5,5)]
#output:- [(9,1),(2,3),(5,4),(5,5),(7,6),(5,9)]

my_list = [(5,4),(9,1),(2,3),(5,9),(7,6),(5,5)]
n = len(my_list)
for i in range(n):
    for j in range(0, n - i - 1):
        if my_list[j][1] > my_list[j + 1][1]:  # compare last elements
            my_list[j], my_list[j + 1] = my_list[j + 1], my_list[j]

print(my_list)
```

```python
#Write a Python program to find the list of words that are longer than or equal to 4 from a given string.
#Input:- 'How much wood would a woodchuck chuck if a woodchuck could chuck wood'
#Output:- ['much', 'wood', 'would', 'woodchuck', 'chuck', 'could']
#Note:- Duplicate should be avoided.

my_str = 'How much wood would a woodchuck chuck if a woodchuck could chuck wood'
result = []
for word in my_str.split():
    if len(word) >= 4 and word not in result:
        result.append(word)

print(result)
```

```python
#Write a Python program which takes two list as input and returns True if they have at least 3 common elements.
#inp_lst1 = [10,20,'Python', 10.20, 10+20j, [10,20,30], (10,20,30)]
#inp_lst2 = [(10,20,30),1,20+3j,100.2, 10+20j, [10,20,30],'Python']
#Output:- True

inp_lst1 = [10,20,'Python', 10.20, 10+20j, [10,20,30], (10,20,30)]
inp_lst2 = [(10,20,30),1,20+3j,100.2, 10+20j, [10,20,30],'Python']
count = 0
for item in inp_lst1:
    if item in inp_lst2:
        count += 1

print(count >= 3)
```

```python
#Write a Python program to create a 4X4 2D matrix with below elements using loop and list comprehension both.
#Output:- [[0,0,0,0],[0,1,2,3],[0,2,4,6],[0,3,6,9]]
matrix = []
for i in range(4):
    row = []
    for j in range(4):
        row.append(i * j)
    matrix.append(row)

print(matrix)

```

```python
#Write a Python program which takes a list of numbers as input and prints a new list after removing even numbers from it.
#Input:- [10,21,22,98,87,45,33,1,2,100]
#Output:- [21,87,45,33,1]
my_list = [10,21,22,98,87,45,33,1,2,100]
result = []
for num in my_list:
    if num % 2 != 0:   # check for odd number
        result.append(num)

print(result)

```
