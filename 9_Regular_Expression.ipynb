```python
#Write the regular expression and use proper method which gives output as:- ['mat', 'man'] in str.
import re
str = 'cat mat bat rat man'
result = re.findall(r'ma.', str)

print(result)
```

```python
#Write the regular expression and use proper method which gives output as:-
#['Python', 's', 'Programming', 'is', 'very', 'easy', 'to', 'learn'] in str.
import re
str = "Python's Programming: is very easy to learn"
result = re.findall(r'\w+', str)

print(result)
```

```python
#Write the regular expression and use proper method which Retrieves all words starting with p.
#Output:- ['peter', 'per', 'picked', 'peck', 'pickled', 'peppers']
import re
str = "peter giper picked a peck of pickled peppers"
result = re.findall(r'\bp\w*', str)

print(result)
```

```python
#Write the regular expression and use proper method which Retrieves all words starting with a digit.
#Output:- ['8th', '11th']
import re
str = 'The election in delhi will be held on 8th and result for the same will be declared on 11th'
result = re.findall(r'\b\d\w*', str)

print(result)
```

```python
#Write the regular expression and use proper method which Retrieves all words having 5 characters.
#Output:- ['peter', 'giper']
import re
str = "peter giper picked a peck of pickled peppers"
result = re.findall(r'\b\w{5}\b', str)

print(result)
```

```python
#Write the regular expression and use proper method which Retrieves a phone number from the given string.
#Output:- 1234567890
import re
str = "Learnbay : 1234567890"
result = re.search(r'\d{10}', str)

print(result.group())
```

```python
##Write the regular expression and use proper method which Etracts name from the string but not number.
#Output:- Learnbay :
import re
str = "Learnbay : 1234567890"
result = re.search(r'[A-Za-z]+\s:', str)

print(result.group())
```

```python
##Write the regular expression and use proper method which Retrieves name starting with 'an' or 'ak'.
#Output:- ['anil', 'akhil', 'anant', 'ankur']
import re
str = 'anil akhil anant abhi arun arati arundhati abhijit ankur'
result = re.findall(r'\b(?:an|ak)\w*', str)

print(result)
```
